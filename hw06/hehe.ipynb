__Задание 1. (0.5 балла)__

Мы будем использовать данные соревнования [Home Credit Default Risk](https://www.kaggle.com/c/home-credit-default-risk/data).  

* Загрузите таблицу **application_train.csv**;
* Запишите в Y столбец с целевой переменной;
* Удалите ненужные столбцы (для этого воспользуйтесь описанием);
* Определите тип столбцов и заполните пропуски - стратегия произвольная;
* Разбейте выборку в соотношении 70:30 с random_state=0.

Так как в данных значительный дисбаланс классов, в качестве метрики качества везде будем использовать площадь под precision-recall кривой.

```python
import numpy as np
import pandas as pd
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)

# Загрузка данных
data = pd.read_csv("application_train.csv")

# Запись целевой переменной
Y = data['TARGET']

# Удаление ненужных столбцов (например, 'SK_ID_CURR')
data = data.drop(columns=['SK_ID_CURR'])

# Вывод информации о столбцах и их типах
data.info()

column_object = []
column_category = []
column_features = []


# Заполнение пропусков (стратегия может быть произвольной)
# К примеру, для числовых значений можно использовать среднее, для категориальных - самый частый элемент.
for column in data.columns:
    if data[column].nunique() < 10:
        data[column] = data[column].astype('category')
        column_category.append(column)
    if data[column].dtype == 'object':
        # Заполнение категориальных пропусков
        data[column].fillna(data[column].mode()[0], inplace=True)
        column_object.append(column)
    elif data[column].dtype != 'category':
        # Заполнение числовых пропусков
        data[column].fillna(data[column].mean(), inplace=True)
        column_features.append(column)

from sklearn.model_selection import train_test_split

# Разделение выборок
X_train_full, X_test_full, y_train, y_test = train_test_split(data, Y, test_size=0.3, random_state=0)

X_train_object, X_train_features = pd.DataFrame(X_train_full[column_object + column_category]), pd.DataFrame(X_train_full[column_features])
X_test_object, X_test_features = pd.DataFrame(X_test_full[column_object + column_category]), pd.DataFrame(X_test_full[column_features])

# X_train_features.info(verbose=True)
# X_train_features.describe(include='all').T
# X_train_features.nunique()
print(np.shape(X_train_features))
X_train_features.head()
```

Также мы будем использовать две реализации градиентного бустинга: [LightGBM](https://lightgbm.readthedocs.io/en/stable/Python-API.html) и [Catboost](https://catboost.ai/en/docs/), которые вам необходимо самостоятельно изучить и установить, используя команды:

Обучите реализации градиентного бустинга LightGBM и Catboost на вещественных признаках без подбора параметров. 
Почему получилась заметная разница в качестве? 

В этом и последующих экспериментах необходимо измерять время обучения моделей.

```python
# # Отбор только вещественных признаков
# X_train_real = X_train.select_dtypes(include=['float64', 'int64'])
# X_test_real = X_test.select_dtypes(include=['float64', 'int64'])

import time
import lightgbm as lgb
from sklearn.metrics import precision_recall_curve, auc

# Инициализация модели
lgb_model = lgb.LGBMClassifier(verbose=0)

# Измерение времени обучения
start_time = time.time()
lgb_model.fit(X_train_features, y_train)
end_time = time.time()

# Оценка времени
lgb_training_time = end_time - start_time

# Предсказания и оценка качества модели
lgb_pred_probs = lgb_model.predict_proba(X_test_features)[:, 1]
precision, recall, _ = precision_recall_curve(y_test, lgb_pred_probs)
lgb_pr_auc = auc(recall, precision)

print("LightGBM - Время обучения:", lgb_training_time)
print("LightGBM - AUC-PR:", lgb_pr_auc)



from catboost import CatBoostClassifier

# Инициализация модели
cat_model = CatBoostClassifier(verbose=False)

# Измерение времени обучения
start_time = time.time()
cat_model.fit(X_train_features, y_train)
end_time = time.time()

# Оценка времени
cat_training_time = end_time - start_time

# Предсказания и оценка качества модели
cat_pred_probs = cat_model.predict_proba(X_test_features)[:, 1]
precision, recall, _ = precision_recall_curve(y_test, cat_pred_probs)
cat_pr_auc = auc(recall, precision)

print("CatBoost - Время обучения:", cat_training_time)
print("CatBoost - AUC-PR:", cat_pr_auc)

```

Почему может быть заметная разница в качестве?
Различие в качестве между LightGBM и CatBoost объясняется рядом факторов:

Обработка категориальных признаков: CatBoost лучше справляется с категориальными признаками, так как использует специальный алгоритм, учитывающий порядки и разбиения. В нашем случае мы использовали только вещественные признаки, но CatBoost может применять более сложные техники даже с числовыми данными.

Алгоритмы построения деревьев: LightGBM использует метод Leaf-wise (по листьям), который быстрее, но склонен к переобучению на малых данных. CatBoost применяет Symmetric Tree и более устойчив к переобучению, что помогает в сложных и дисбалансированных задачах.

Обработка дисбаланса классов: CatBoost имеет встроенные методы для работы с дисбалансом классов, что улучшает качество предсказания для менее представленных классов.

Скорость обучения и оптимизация: LightGBM известен своей высокой скоростью благодаря оптимизации построения деревьев и использованию histogram-based подхода. Однако CatBoost может показать лучшие результаты по качеству, так как он применяет особые техники сглаживания и обработки редких значений.

Эти факторы объясняют разницу в показателях качества и скорости обучения между моделями.

Подготовка параметров для подбора

Чтобы определить диапазоны для параметров, создадим сетки гиперпараметров:

* Глубина деревьев (max_depth): определяет, насколько глубокими будут деревья. Более глубокие деревья могут лучше соответствовать данным, но склонны к переобучению.
* Количество деревьев (n_estimators): число деревьев в ансамбле.
* Темп обучения (learning_rate): указывает, с какой скоростью модель обновляет веса для каждого дерева. Низкие значения темпа часто компенсируются большим количеством деревьев.
* Оптимизируемый функционал (objective): функция потерь, используемая при обучении модели.

```python
import lightgbm as lgb
from sklearn.model_selection import GridSearchCV

X_train_temp = X_train_features
y_train_temp = y_train



# Параметры для подбора
param_grid_lgb = {
    'max_depth': [3, 5, 7],
    'n_estimators': [2, 3, 4, 5],
    'learning_rate': [0.01, 0.05, 0.1],
    'objective': ['binary', 'cross_entropy']  # Пример для задачи бинарной классификации
}

# Инициализация модели
lgb_model = lgb.LGBMClassifier()

# Кросс-валидация с GridSearchCV
grid_search_lgb = GridSearchCV(estimator=lgb_model, param_grid=param_grid_lgb, cv=3, scoring='average_precision', n_jobs=-1, verbose=1)
grid_search_lgb.fit(X_train_temp, y_train_temp)

# Лучшие параметры и результаты
best_params_lgb = grid_search_lgb.best_params_
best_score_lgb = grid_search_lgb.best_score_

print("Лучшие параметры для LightGBM:", best_params_lgb)
print("Лучшее качество для LightGBM (AUC-PR):", best_score_lgb)

# В данной конфигурации lgb подбирается примерно за минуту+



from catboost import CatBoostClassifier

# Параметры для подбора
param_grid_cat = {
    'depth': [3, 5, 7],
    'iterations': [2, 3, 4, 5],
    'learning_rate': [0.01, 0.05, 0.1],
    'loss_function': ['Logloss', 'CrossEntropy']  # Пример для бинарной классификации
}

# Инициализация модели
cat_model = CatBoostClassifier(verbose=0)

# Кросс-валидация с GridSearchCV
grid_search_cat = GridSearchCV(estimator=cat_model, param_grid=param_grid_cat, cv=3, scoring='average_precision', n_jobs=-1, verbose=1)
grid_search_cat.fit(X_train_temp, y_train_temp)

# Лучшие параметры и результаты
best_params_cat = grid_search_cat.best_params_
best_score_cat = grid_search_cat.best_score_

print("Лучшие параметры для CatBoost:", best_params_cat)
print("Лучшее качество для CatBoost (AUC-PR):", best_score_cat)

# В данной конфигурации CatBoostClassifier подбирается примерно за пол минуты
```

Анализ соотношения глубины и количест

После выполнения подбора параметров можно проанализировать, как оптимальные значения глубины деревьев и количества деревьев отличаются между LightGBM и CatBoost:

* LightGBM: Часто лучше работает с небольшими значениями глубины деревьев, поскольку его метод построения Leaf-wise с большей глубиной склонен к переобучению. Однако с увеличением n_estimators LightGBM может достичь высоких результатов, используя более мелкие деревья.
* CatBoost: Может эффективно использовать более глубокие деревья благодаря симметричным разбиениям, что снижает риск переобучения и помогает лучше справляться с дисбалансом. Однако для увеличения точности (эквивалент n_estimators) также имеет значение.
